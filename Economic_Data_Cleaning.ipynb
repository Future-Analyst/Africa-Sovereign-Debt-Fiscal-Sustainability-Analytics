## Create a Function that converts Amount column to Apporpriate Value

```python
def convert_to_usd(df):
    """
    Converts monetary indicators to USD million while preserving
    non-monetary indicators in their original units.

    Handles mixed million/billion monetary units.

    Required columns:
    Country, Indicator, Amount, Unit
    """

    exchange_rates = {
        "Egypt": 49.38,
        "Ethiopia": 217.30,
        "Ghana": 11.29,
        "Ivory Coast": 578.00,
        "Nigeria": 1380.43,
        "South Africa": 22.26,
        "Tanzania": 3571.00,
        "Togo": 578.00,
        "Senegal": 578.00,
        "Kenya": 129.60,
        "Rwanda": 2000.00,
        "Algeria": 133.23,
        "Angola": 1250.00,
        "Botswana": 13.40
    }

    df = df.copy()

    # Clean text columns
    df["Indicator"] = df["Indicator"].astype(str).str.strip()
    df["Unit"] = df["Unit"].astype(str).str.strip()
    df["Country"] = df["Country"].astype(str).str.strip()


    # Indicators measured as monetary values
    monetary = {
        "Budget Deficit",
        "Capital Expenditure",
        "Defence Expenditure",
        "Education Expenditure",
        "Expenditure",
        "Exports",
        "GDP per Capita",
        "Government Debt",
        "Health Expenditure",
        "Imports",
        "Nominal GDP",
        "Real GDP",
        "Revenue",
        "Tax Revenue",
        "Value Added Tax (VAT)"
    }


    percentage = {
        "Inflation Rate",
        "Food Inflation",
        "Food Inflation YoY",
        "GDP Growth Rate",
        "Interest Rate",
        "Unemployment Rate"
    }


    index = {
        "Consumer Price Index (CPI)"
    }


    count = {
        "Population",
        "Labour Force"
    }


    # Start with original amount
    df["Value"] = df["Amount"]


    # Monetary rows
    monetary_mask = df["Indicator"].isin(monetary)


    # Convert billions into millions
    billion_mask = (
        monetary_mask &
        df["Unit"].str.contains(
            r"Billion|Billions|bn|B",
            case=False,
            na=False
        )
    )


    df.loc[billion_mask, "Value"] = (
        df.loc[billion_mask, "Value"] * 1000
    )


    # Convert local currency to USD
    df.loc[monetary_mask, "Value"] = (
        df.loc[monetary_mask, "Value"] /
        df.loc[monetary_mask, "Country"].map(exchange_rates)
    )


    # Final unit labels
    df.loc[monetary_mask, "Unit"] = "USD million"

    df.loc[df["Indicator"].isin(percentage), "Unit"] = "%"

    df.loc[df["Indicator"].isin(index), "Unit"] = "Index"

    df.loc[df["Indicator"].isin(count), "Unit"] = "Persons"


    return df
```

```python
# convert the amount column to numeric
df5["Amount"] = pd.to_numeric(df5["Amount"], errors='coerce')
converted_df = convert_to_usd(df5)
```
